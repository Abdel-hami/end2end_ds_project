```python
API_URL = 'http://127.0.0.1:5000/'
PREDICTION_ENDPOINT = 'predict_jobs_probs'
```

```python
import requests
import json
import pandas as pd
```

## Predict

```python
available_skills = ['python','Keras', 'Pandas','TensorFlow', 'Torch/PyTorch']
```

```python
response = requests.post(API_URL + PREDICTION_ENDPOINT,
                           json = available_skills)

response.raise_for_status()

predictions = response.json()

pd.Series(predictions).sort_values(ascending=False)
```
